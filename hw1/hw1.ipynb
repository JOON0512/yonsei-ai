#### P2.3 Implement the ```learning_with_gradient_descent``` function in the ```LinearRegressor``` class, and describe your implementation. Refer to the algorithm in the lecture note. (5 point)

<span style="color: yellow;">

#### P2.3 Answer

Gradient Descent is an iterative optimization algorithm used to find the minimum of a function. In the context of linear regression, we use it to minimize the Residual Sum of Squares (RSS) loss function by adjusting the weights incrementally.

##### Convergence Condition
To determine convergence, we stop the iteration when the change in the weight vector \( w \) is smaller than a specified tolerance \( \tau \). In this case, the condition is:

\[
||\hat{w}_t - \hat{w}_{t+1}||_\infty < \tau
\]

where:
- \( \hat{w}_t \) represents the estimated weights at iteration \( t \),
- \( \tau = 10^{-3} \) is the convergence threshold,
- The infinity norm \( ||x||_\infty \) of a vector \( x = (x_1, x_2, \ldots, x_n) \) is defined as:

  \[
  ||x||_\infty = \max(|x_1|, |x_2|, \ldots, |x_n|)
  \]


Initialization: The function initializes the weights weight_gradient_descent using a normal distribution (with initialize_weight()) and calculates the initial RSS loss, which is stored in self.losses. This initialization provides a starting point for the gradient descent optimization process.

Gradient Computation: In each iteration, the gradient of the RSS (Residual Sum of Squares) loss with respect to each weight is computed. This gradient is calculated as:

gradient=−2×𝑋𝑇×(𝑦−𝑋⋅current_w)

where X is the input data, y is the target data, and current_w is the current weight vector.

Weight Update: The weights are then updated by moving in the opposite direction of the gradient. The learning rate lr controls the step size:

current_w=prev_w−lr×gradient

This update rule allows the model to minimize the RSS loss iteratively.

Convergence Check: The convergence criterion checks if the maximum absolute difference between prev_w and current_w is less than the threshold tau. If this condition is met, the gradient descent is considered to have converged, and the loop breaks. The final weights are saved as weight_gradient_descent.

Recording Loss History: After each iteration, the current RSS loss is computed and appended to self.losses, allowing us to monitor the loss reduction over iterations.

````python
time_start = time.time()
linear.learning_with_gradient_descent(X, y)
print(f"Time elapsed: {time.time() - time_start}")
print(f"Final loss: {linear.losses[-1]}")
linear.loss_history()
````

P2.4 Implement the ```learning_with_closed_form``` function in the ```LinearRegressor``` class, and describe your implementation. (5 point)

<span style="color: yellow;">

#### P2.4 Answer
<span style="color: yellow;">

In linear regression, the closed-form solution provides a direct way to calculate the optimal weights that minimize the sum of squared errors between predicted and actual values. This approach eliminates the need for iterative methods, such as gradient descent, by using a mathematical formula derived from the least squares criterion.

##### Derivation
Given a dataset with input data \( X \) and target values \( y \), we aim to model the relationship between \( X \) and \( y \) using a linear model:
\[
y = Xw + \epsilon
\]
where:
- \( w \) is the vector of weights (or coefficients) we wish to determine,
- \( \epsilon \) represents the error term.

To find the optimal \( w \) that minimizes the residual sum of squares (RSS), we solve the following optimization problem:
\[
\min_w ||Xw - y||^2
\]
Taking the derivative with respect to \( w \) and setting it to zero leads to:
\[
X^TXw = X^Ty
\]
This results in the closed-form solution:
\[
w = (X^TX)^{-1}X^Ty
\]

Matrix Transpose: The code first computes the transpose of the matrix X, storing it in X_transpose. This transpose will be used in the formula to compute the closed-form solution for linear regression.

Closed-Form Solution: The optimal weights for linear regression can be computed using the normal equation:

𝑤=(𝑋.𝑇 𝑋)^−1 𝑋.𝑇 𝑦

Here:

X is the input matrix containing all data points.
y is the target vector.
w represents the weight vector (coefficients) that minimizes the residual sum of squares.
Matrix Inversion and Multiplication:

np.linalg.inv(X_transpose @ X) computes the inverse of the matrix 𝑋.𝑇 𝑋.
The inversion is essential to solve for w analytically in the absence of regularization.
@ denotes matrix multiplication in Python.
X_transpose @ y computes the dot product of the transposed X matrix and the target vector y, which gives us the necessary component for the solution.
Set the Result: The computed weights are assigned to self.weight_closed_form. This attribute now contains the optimal weights for predicting y from X based on the closed-form solution.

````python
time_start = time.time()
linear.learning_with_closed_form(X, y)
print(f"Time elapsed: {time.time() - time_start}")
print(f"Final loss: {RSSloss(X, y, linear.weight_closed_form)}")
````

### P5. Elastic Regression (10 point)
#### P5.1 Implement the ```ElasticRegressor``` class, and describe your implementation. (10 point)
#### Note that **you should write the derivation process of threshold in your answer.** 

$$
\mathrm{Total~cost}=RSS(w)+\lambda\left( \alpha\sum_j\lvert w_j \rvert + (1-\alpha)\sum_j w_j^2 \right)
$$

<span style="color: yellow;">

#### P5.1 Answer

The Elastic Net objective function combines the residual sum of squares (RSS) with both ℓ1(absolute) and ℓ2(squared) penalties:

Total cost = 𝑅𝑆𝑆(𝑤) + 𝜆(𝛼∑𝑗 ∣ 𝑤_𝑗 ∣ + (1−𝛼)∑𝑗 𝑤_𝑗 ^2)

where:

λα∑j ∣wj∣ applies an ℓ1 penalty (Lasso) which can set some coefficients to zero.

λ(1−α)∑j (w_j)^2 applies an ℓ2 penalty (Ridge) that shrinks coefficients but doesn’t drive them exactly to zero.

Elastic Net uses a coordinate descent approach, where we update each weight wj individually. To do this, we derive the update rule for each wj by setting the partial derivative of the cost function with respect to wj to zero. The derivation process for the threshold in Elastic Net regression stems from minimizing the cost function with respect to each weight wj separately in a coordinate descent approach. The Elastic Net regularization combines ℓ1(Lasso) and ℓ2(Ridge) penalties, each influencing the threshold for each weight update.

-Class Initialization:
    ElasticNetRegressor class combines Lasso and Ridge regularization.
    It initializes parameters dim (feature dimension), tau (convergence threshold), lam (regularization parameter), and alpha (mixing parameter for Lasso and Ridge).
    Here, alpha = 1 indicates pure Lasso, while alpha = 0 represents pure Ridge. Any value in between balances both regularizations.

-Thresholding Function:
    The thresholding function is used for Lasso regularization to enforce sparsity.
    It calculates a threshold value based on the parameter 𝛼𝜆/2. If 𝜌(the residual) is less than −𝛼𝜆/2​, it adds 𝛼𝜆/2 to rho; if ρ is greater than 𝛼𝜆/2, it subtracts 𝛼𝜆/2 from rho. Otherwise, it sets rho to zero. This soft-thresholding ensures some coefficients shrink to zero.

-Learning with Coordinate Descent:
Objective Function: The total cost is given by:

Total cost = 𝑅𝑆𝑆(𝑤) + 𝜆(𝛼∑𝑗 ∣𝑤j∣ + (1−𝛼)∑𝑗 (𝑤𝑗) ^2)

Objective: Minimize Each wj Separately:
To minimize the cost function for each weight wj, we look at the partial derivative of the cost function with respect to wj and set it to zero.
​
For simplicity, let's focus on the terms involving wj in the cost function:
f(wj) = 𝑅𝑆𝑆(𝑤j) + 𝜆(𝛼∑𝑗 ∣𝑤𝑗∣ + (1−𝛼)∑𝑗 (𝑤𝑗) ^2)

Let:
ρj = i∑ Xij(yi − ∑k!=j Xik * wk)

Then, the cost function in terms of wj becomes

​f(wj)=−2ρjwj + wj^2 + λ(α∣wj∣ + (1−α)wj^2)

Taking the derivative of f(wj) with respect to wj:

For the ℓ1 term (𝛼∣𝑤𝑗∣): This term introduces a constant threshold proportional to λα.
For the ℓ2 term ((1−𝛼)𝑤𝑗^2): This term scales linearly with wj and contributes 𝜆(1−𝛼)𝑤𝑗.

#### P6.4 Explain and compare three regression methods (Ridge, Lasso, and Elastic Net). (10 point)

<span style="color: yellow;">

#### P6.4 Answer

Ridge provides stability and efficiency with a closed-form solution (under certain conditions) but lacks feature selection. Lasso offers feature selection by setting some coefficients to zero, making the model simpler and more interpretable, though it requires iterative methods to solve. Elastic Net combines the benefits of Ridge and Lasso, making it effective in high-dimensional data with correlated features, but it also requires iterative optimization due to the L1 component.

Ridge Regression:

Unlike Lasso, Ridge regression has a differentiable regularization term, allowing a closed-form solution (i.e., weights can be solved in a single step when the matrix inverse exists). This is an advantage because it can be computationally efficient, especially in scenarios where finding an exact solution is preferable. However, a closed-form solution may not be feasible if the matrix is singular or not invertible. From the project results, we observe that when the regularization parameter λ is large, Ridge tends to have a smaller error than Lasso. This is because Ridge regression shrinks all feature weights proportionally, which prevents any one feature from dominating, leading to a more stable model. A limitation of Ridge regression is its inability to perform feature selection. Since it only shrinks coefficients but does not set them to zero, every feature remains in the model, contributing to predictions, which can make interpretation more challenging.

Lasso Regression:

A major advantage of Lasso regression is its ability to perform feature selection. By driving the weights of less influential features to zero, Lasso creates a sparse model that includes only the most relevant features. This makes Lasso particularly useful when we suspect that only a few features are significant predictors. Another observation from the results is that with a smaller λ, Lasso often produces a smaller error than Ridge, as Lasso’s weights converge quickly to zero, eliminating noise and irrelevant features. However, as λ increases too much, Lasso may eliminate too many features, potentially reducing model accuracy. One drawback of Lasso is that its regularization term is not differentiable at wj = 0. This means that a closed-form solution does not exist, and we must rely on iterative optimization methods to find the solution, which can be computationally more intensive.

Elastic Net Regression:

Elastic Net combines the advantages of both Ridge and Lasso by including both L1 (Lasso) and L2 (Ridge) regularization terms. The balance between these two is controlled by a mixing parameter α, allowing Elastic Net to achieve both feature selection and weight shrinkage. Elastic Net is particularly useful when dealing with high-dimensional datasets where features may be correlated or collinear. Ridge alone may struggle with multicollinearity, while Lasso might select only one feature from a group of correlated features. Elastic Net’s combination of L1 and L2 penalties can select multiple correlated features while also shrinking their magnitudes, leading to better generalization. Unlike Lasso, Elastic Net can handle scenarios where there are more features than observations and where collinearity exists. However, like Lasso, Elastic Net requires iterative optimization due to the non-differentiability of the L1 component.

#### P6.5 Observe training and test errors over the value of $\lambda$. Discuss the effect of the tuning parameter $\lambda$. (10 point)

<span style="color: yellow;">

#### P6.5 Answer

Increasing λ generally leads to higher training and test errors due to reduced model flexibility and complexity. This effect highlights the trade-off between bias and variance:
Low λ: The model is more flexible and may overfit the training data, yielding low training error but potentially high test error if overfitting occurs.
High λ: The model complexity decreases, reducing the risk of overfitting, but excessive penalization can lead to underfitting, causing both training and test errors to increase.
In practical applications, an optimal λ balances this trade-off by minimizing test error without incurring high training error.

Least Squares:

As expected, the Least Squares method, which lacks a regularization term, shows consistent training and test errors across all values of λ, as λ has no impact on its formulation.

Ridge Regression:

Ridge regression demonstrates a relatively steady increase in both training and test errors as λ increases. This is because, with higher λ, the regularization term penalizes larger weights, resulting in lower model complexity and, consequently, a less flexible model. At lower values of λ, Ridge regression minimizes training error effectively, but as λ increases, the error begins to rise since the model becomes more regularized and less fitted to the data. This steady rise in error reflects Ridge’s smooth and gradual weight decay, unlike Lasso.

Lasso Regression:

Lasso regression has a noticeable increase in both training and test errors as λ increases, and this increase is more pronounced than in Ridge. Lasso’s regularization term leads to weights being set to zero for features with little impact, effectively performing feature selection. As λ grows, Lasso quickly penalizes the weights, leading to a sparse solution with fewer features contributing to the model. This aggressive penalization can lead to a sudden increase in error, as shown in the training and test error curves, which grow more sharply than Ridge’s curves.

Elastic Net Regression:

Elastic Net, a combination of Ridge and Lasso, balances the smooth decay in weights from Ridge with the sparsity-inducing effect of Lasso. As a result, the training and test error curves increase with λ but are typically positioned between the Ridge and Lasso curves. Elastic Net leverages both types of regularization to prevent overfitting while maintaining model interpretability, offering a middle ground in terms of error increase rates. The gradual slope of the error curves is a result of this combined approach, making Elastic Net less sensitive to λ than Lasso but still effective in reducing model complexity.

````python
def stack_weight_over_lambda(X, y, model_type, tau, dim, lambda_list):
    stacked_weight = np.zeros([len(lambda_list), X.shape[1]])
    for i in range(len(lambda_list)):
        model = eval(model_type + 'Regressor')(tau=tau, dim=dim, lam=lambda_list[i])
        model.learning_with_coordinate_descent(X, y)
        stacked_weight[i, :] = model.weight_coordinate_descent
    return stacked_weight

def compute_errors(X, y, lambda_list, weights):
    assert len(lambda_list) == len(weights)
    rss_errors = []
    for i in range(len(lambda_list)):
        error = RSSloss(X, y, weights[i, :])
        rss_errors.append(error)
    return rss_errors
````

````python
lambda_list = [i for i in range(0, 201, 5)]
print(lambda_list)
````

````python
stacked_weight_ridge = stack_weight_over_lambda(X, y, 'Ridge', 1e-3, X.shape[1], lambda_list)
stacked_weight_lasso = stack_weight_over_lambda(X, y, 'Lasso', 1e-3, X.shape[1], lambda_list)
stacked_weight_elsatic = stack_weight_over_lambda(X, y, 'ElasticNet', 1e-3, X.shape[1], lambda_list)
````

````python
training_errors_ridge = compute_errors(X, y, lambda_list, stacked_weight_ridge)
training_errors_lasso = compute_errors(X, y, lambda_list, stacked_weight_lasso)
training_errors_elastic = compute_errors(X, y, lambda_list, stacked_weight_elsatic)
````

````python
plt.figure(figsize=[15, 8])
plt.plot(lambda_list, [training_errors_lasso[0] for _ in range(len(training_errors_lasso))], label='Least Squares', linestyle='--', color='lightgrey')
plt.plot(lambda_list, training_errors_ridge, label='Ridge')
plt.plot(lambda_list, training_errors_lasso, label='Lasso')
plt.plot(lambda_list, training_errors_elastic, label='Elastic Net')
plt.title(f"Training error over $\lambda$", fontsize=30)
plt.xlabel('$\lambda$', fontsize=20)
plt.ylabel('Error', fontsize=20, rotation=90)
plt.legend(fontsize=20)
plt.show()
````
